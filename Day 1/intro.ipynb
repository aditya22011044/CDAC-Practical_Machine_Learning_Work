```python
age = [20, 30, 45, 60, 70, 50]
```

```python
sum = 0
for i in age:
    sum+=i
mean = sum/len(age)
print(mean)
```

```python
import math
y = 0
for i in age:
    y += (i-mean)**2

s = math.sqrt(y/5)
print(s)
```

```python
std = []
ss = 0
for i in age:
    std.append((i-mean)/s)
    ss+=(i-mean)/s

print(std)
print(ss/5)
```

```python
a=0
ssd = 0
for i in std:
    a += (i-mean)**2

ssd = math.sqrt(a/5)
print(ssd)
```
